This notebook inspects the curated PEER NGA-West2 earthquake records, baseline correction protocols, and response spectra across magnitude-distance-Vs30 bins.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
from src.data_pipeline.ground_motion_loader import load_processed_ground_motion

# Load dataset metadata index
df = pd.read_csv("data/simulations/simulation_index.csv")
print(f"Total simulations in database: {len(df):,}")
print(f"Unique earthquakes: {df['earthquake_name'].nunique()}")
print(f"PGA scale factors (g): {sorted(df['target_pga_g'].unique())}")
df.head(5)
```

```python
# Plot PGA and Ductility Distributions
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].hist(df["target_pga_g"], bins=10, color="#3B82F6", edgecolor="black", alpha=0.8)
axes[0].set_xlabel("Target PGA (g)")
axes[0].set_ylabel("Count")
axes[0].set_title("Ground Motion Scaling Distribution")
axes[0].grid(True, linestyle="--", alpha=0.6)

axes[1].hist(df["ductility"], bins=30, color="#10B981", edgecolor="black", alpha=0.8)
axes[1].set_xlabel("Ductility Demand $\mu = u_{\max} / u_y$")
axes[1].set_ylabel("Count")
axes[1].set_title("Structural Inelastic Ductility Distribution")
axes[1].grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```
