# 02. MDOF Ground Truth Verification
**Multi-Degree-of-Freedom Modal Validation & Rayleigh Damping**

Validates the N-story shear building model against closed-form modal eigenvalue frequencies.


```python
import numpy as np
from src.ground_truth.opensees_mdof_model import MDOFShearBuilding, MDOFParams

# Initialize 3-story shear building
params_3s = MDOFParams(n_stories=3, floor_mass=1000.0, story_stiffness=4e5, zeta_1=0.05, zeta_2=0.05)
model_3s = MDOFShearBuilding(params_3s)

# Compute closed-form analytical modal frequencies
omega_modal, periods_modal = model_3s.compute_modal_properties()
print("3-Story Natural Periods (Closed-Form):")
for i, T in enumerate(periods_modal):
    print(f"  Mode {i+1}: T = {T:.4f} s (f = {1/T:.2f} Hz)")
```

```python
# Run NLTHA under earthquake motion
dt = 0.01
ag = 0.3 * 9.80665 * np.random.randn(1000)
res = model_3s.simulate_nltha(ag, dt=dt)
print("NLTHA Execution Successful:")
print(f"  Displacement shape: {res['drift'].shape} (stories x time)")
print(f"  Max top floor displacement: {np.max(np.abs(res['drift'][-1])) * 1000:.2f} mm")
```
