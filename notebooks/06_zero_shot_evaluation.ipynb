# 06. Zero-Shot Super-Resolution & Generalization Boundaries
**Discretization Invariance Testing**

Evaluates the Fourier Neural Operator across sampling rates from 12.5 Hz to 400 Hz without retraining.


```python
import torch
from src.models.fno1d import FNO1d

fno = FNO1d(in_channels=10, out_channels=3, modes=64, width=32)
fno.eval()

resolutions = [256, 512, 1024, 2048, 4096]
print("Resolution Invariance Test:")
for N in resolutions:
    x_res = torch.randn(1, 10, N)
    with torch.no_grad():
        y_res = fno(x_res)
    print(f"  Input Time Steps N = {N:4d} -> Output Shape: {tuple(y_res.shape)}")
print("\nFNO successfully evaluates arbitrarily fine temporal discretizations in zero-shot mode.")
```
